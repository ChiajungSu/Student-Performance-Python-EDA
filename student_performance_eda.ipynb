# Student Performance: Exploratory Data Analysis

STAT 5606 Final Project. Dataset source: [Kaggle – Student Performance](https://www.kaggle.com/datasets/neurocipher/student-performance/data).

Run cells in order with `StudentPerformance.csv` in the working directory.


```python
# 1) Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
FILE_NAME = "StudentPerformance.csv"
df = pd.read_csv(FILE_NAME)

print("Loaded:", FILE_NAME)
print("Shape (rows, cols):", df.shape)
```

```python
# clean data
df.columns = df.columns.str.strip()
print("\nColumns:")
print(list(df.columns))
```

```python
print("\nInfo:")
df.info()

print("\nNumeric describe:")
print(df.select_dtypes(include=np.number).describe())
```

```python
# Missing value
print("\nMissing values per column:")
print(df.isnull().sum().sort_values(ascending=False))
```

```python
# Duplicates
dup_count = df.duplicated().sum()
print("\nDuplicated rows:", dup_count)
if dup_count > 0:
    df = df.drop_duplicates()
    print("Duplicates removed. New shape:", df.shape)
```

```python
cat_cols = df.select_dtypes(include=["object"]).columns
df[cat_cols] = df[cat_cols].astype("category")

print("\nCategorical columns converted to 'category':")
print(list(cat_cols))
```

```python
# EDA
target_candidates = ["Performance Index", "performance index", "Performance", "Score", "score"]
target_col = None
for tc in target_candidates:
    for col in df.columns:
        if col.lower() == tc.lower():
            target_col = col
            break
    if target_col:
        break

print("\nTarget column detected:", target_col)
```

```python
# Distribution
if target_col:
    plt.figure(figsize=(8,5))
    plt.hist(df[target_col], bins=20)
    plt.title(f"Distribution of {target_col}")
    plt.xlabel(target_col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
# Export
OUTPUT_NAME = "student_performance_cleaned.csv"
df.to_csv(OUTPUT_NAME, index=False)
print(f"\n Exported cleaned dataset: {OUTPUT_NAME}")

# (Colab) Optional: download the exported file automatically
try:
    from google.colab import files
    files.download(OUTPUT_NAME)
except Exception:
    pass
```
